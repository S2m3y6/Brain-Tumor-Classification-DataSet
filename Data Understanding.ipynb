```python
# all the imports
import torchvision.transforms as transf
from torchvision.datasets import ImageFolder as ImageF 
from torch.utils.data import DataLoader
```

```python
# 1. Definieer transformaties (bijv. formaat aanpassen en omzetten naar Tensor)
transform = transforms.Compose([
    transforms.Resize((224, 224)), # Pas het formaat aan naar wens
    transforms.ToTensor()          # Zet de afbeelding om naar een PyTorch Tensor
])

# 2. Laad de dataset in met ImageFolder
# Vervang 'pad/naar/mijn_data' door de map waar de submappen in staan
dataset = ImageFolder(root='./Training', transform=transform)

# 3. Stop de dataset in een DataLoader voor training/testing
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

# Testen of het werkt:
print(f"Aantal klassen gevonden: {len(dataset.classes)}")
print(f"Klassennamen: {dataset.classes}")
print(f"Totaal aantal foto's: {len(dataset)}")
```
